```python
%%time
import matplotlib.pyplot as plt
import numpy as np
from scipy.linalg import eigvals

def f(z, t1, t2):
    """
    z is complex? t1,t2 on the complex unit circle (profconradi.com)
    """
    o = -0.7656*z**2 + 64*z**14 - 84*z**10 + (t2*(-9610 + 1080j)+920+180j)*z**6
    o += (t1**5*(-625+249j) + t1**4*(1520+1140j) + t1**3*(482-2070j) + t1**2*(1790+1210j) + t1*(-2800 + 109j) + 2760+2800j)*z**5
    return o

def plot_points(x, y, fig=None, ax=None, line=None):
    if fig is None and ax is None:
        fig,ax = plt.subplots(figsize=(16,16), dpi=300, facecolor="k")
    #ax.scatter(x, y, s=(72./fig.dpi)**2, marker='o', facecolor='white', lw=0, alpha=0.2) #1 pixel marker
    ax.scatter(x, y, s=1, marker='o', facecolor='white', lw=0, alpha=0.9) #1 pixel marker
    ax.set(facecolor="k")
    ax.set_aspect("equal")
    ax.set(xlim=(-1,1), ylim=(-1,1))
    ax.axis('off')
    fig.tight_layout()
    return fig,ax

Dz = 5
Nz = 100
for Nt in [10]: #, 100000, 1000000]: #, 10000000]:
    print(f"Nz={Nz:d} Nt={Nt:d}...")
    
    fig,ax=None,None
    try:
        xs, ys = np.loadtxt(f"my_bubbly_Nz{Nz}_Nt{Nt}.dat", unpack=True)
    except:
        xs,ys = [],[]
        zs = np.linspace(-Dz,Dz,Nz)
        angles = np.linspace(0,2*np.pi,Nt)
        ts = np.exp(angles*1j)
        for i in range(Nz):
            print(f"\r [ Scanned {i+1}/{Nz} rows of the image... ]", end='')
            for j in range(Nz):
                z = zs[i] + zs[j]*1j
                for k in range(Nt):
                    t1 = ts[k]
                    for l in range(Nt):
                        t2 = ts[l]
                        z_ = f(z, t1, t2)
                        xs.append(z_.real)
                        ys.append(z_.imag)
        print("\rDone                                        ")
        np.savetxt(f"my_bubbly_Nz{Nz}_Nt{Nt}.dat", np.array([xs,ys]).T)
    fig,ax = plot_points(xs, ys, fig=fig, ax=ax)
    for fmt in ["png"]:
        fig.savefig(f"my_bubbly_Nz{Nz}_Nt{Nt}.{fmt}")
    #plt.show()
```
