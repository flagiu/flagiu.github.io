```python
%%time
import matplotlib.pyplot as plt
import numpy as np

def plot_points(x, y, colors, fig=None, ax=None, line=None):
    if fig is None and ax is None:
        fig,ax = plt.subplots(figsize=(16,16), dpi=300, facecolor="k")
    #ax.scatter(x, y, s=(72./fig.dpi)**2, marker='o', facecolor='white', lw=0, alpha=0.2) #1 pixel marker
    colors[colors>0] = 1-colors[colors>0] # invert palette for c>0
    ax.scatter(x, y, c=colors, s=1, marker='o', lw=0, alpha=0.9) #1 pixel marker
    ax.set(facecolor="k")
    ax.set_aspect("equal")
    ax.set(xlim=(min(x),max(x)), ylim=(min(y),max(y)))
    ax.axis('off')
    fig.tight_layout()
    return fig,ax

def get_julia_set(f, zs, R, max_iter, outname_prefix):
    try:
        xs, ys, cs = np.loadtxt(f"{outname_prefix}.dat", unpack=True)
    except:
        xs,ys,cs = [],[],[]
        for i in range(Nz):
            if i%10==9:
                print(f"\r [ Scanned {i+1}/{Nz} rows of the image... ]", end='')
            for j in range(Nz):
                z = zs[i] + zs[j]*1j
                xs.append(z.real)
                ys.append(z.imag)
                iteration = 0
                while np.abs(z)<R and iteration<max_iter:
                    z = f(z)
                    iteration += 1
                if iteration >= max_iter:
                    color = 0.0
                else:
                    color = iteration/max_iter
                cs.append(color)
        print("\rDone computation                                        ")
    return xs, ys, cs

Dz = 3
max_iter = 10000
R = np.sqrt(2)*Dz
def f_conradi(z): # profconradi.com
    return 1/( (0.15+0.15j)*z**5 + z**3 - (3.00-3.00j)*z )

for Nz in [1000]: #, 500, 1000, 2000, 4000]:
    outname_prefix = f"julia_conradi_Dz{Dz:.2f}_Nz{Nz}_maxiter{max_iter}"
    zs = np.linspace(-Dz,Dz,Nz)
    xs, ys, cs = get_julia_set(f_conradi, zs, R, max_iter, outname_prefix)
    np.savetxt(f"{outname_prefix}.dat", np.array([xs,ys,cs]).T)
    print("Done saving data")
    fig,ax = plot_points(xs, ys, cs, fig=None, ax=None)
    print("Done plotting")
    for fmt in ["png"]:
        fig.savefig(f"{outname_prefix}.{fmt}")
        print(f"Done saving {fmt}")
    del fig,ax
```
